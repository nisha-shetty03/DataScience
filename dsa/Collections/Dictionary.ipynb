Dict functions and methods

```python
student={
    'id':"24cs093",
    'name':"Nisha",
    'age':20,
    'dep':"cse"
}

# Retrive a single value using key
getval=student.get('age')
print(getval)

# checking all the keys
dict_key=student.keys()
print(dict_key)

# checking all the values
dict_val=student.values()
print(dict_val)

#updatig values 
update_val=student.update({"age":21})
print(student)

#set default
default_val=student.setdefault('place',"udupi")
print(student)

# copy
c=student.copy()
print(c)

# pop. remove key-value based on key 
student.pop("dep")
print(student)

#pop item - removes least recently added item 
student.popitem()
print(student)

#in oder to use list-of-tuple format
print(student.items())

```

Iterations

```python
student={
    'id':"24cs093",
    'name':"Nisha",
    'age':20,
    'dep':"cse"
}

# iterating over keys
for key in student:
    print(key)

# iterating over values
for val in student.values():
    print(val)

# iterating over key and values
for k,v in student.items():
    print(k,v)
```
