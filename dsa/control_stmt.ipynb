```python
# for loop
# printing even numbers
n=10
for i in range (1,n+1,1):  # syntax: for var in range(start,stop,step):
    print(i,end=" ")
```

```python
# while loop
n=1
while(n!=10):
    print(n,end=" ")
    n+=1
```

```python
#match
bal=10000
while(True):
    print("Enter 1 for deposit")
    print("Enter 2 for withdrawal")
    print("Eneter 3 to check balance")
    print("Enter 4 to exit")
    choice=int(input("Enter your choice"))
    match choice:
        case 1:
            dep=float(input("Enter the amount"))
            bal+=dep
            print("Balance after deposit is:",bal)
        case 2:
            withdraw=float(input("Enter the amount to be withdrawed"))
            if(withdraw<=bal):
                bal-=withdraw
                print("Amount after withdrawal is:",bal)
            else:
                print("Insufficient balance")
        case 3:
            print("Balance is:",bal)
        case 4:
            print("Exiting")
            break
```
