List

```python
# types of list
# list on numbers
nums=[1,2,3,4,5]
# list on strings
str=["red","black","white"]
# list on mixed data type
s=["raju",20,1.9,True]
#  nested list 
s2=[
    [1,2,3],
    [2,3,4],
    [3,4,5]
]
print(nums,"\n",str,"\n",s,"\n",s2)
```

```python
# List functions
colors=["red","blue","black"]
print(len(colors))
colors.append("gold")#inserting an element at the end 
print(colors)
colors[0]="silver" #replacing an element
colors.insert(2,"navy blue") #inserting value at a perticular index
colors.remove("blue") # removing an element
colors.pop()        # removing the last element
colors.append("gold")
print(colors)
```

```python
# Looping 
colors=["red","blue","black"]
for i in range(len(colors)):  #accessing elements using loops 
    print(i,colors[i])
for c in colors:               #accessing the elements directly 
    print(c,end=" ")
```

```python
marks=[98,96,90]
print("Maximum is:",max(marks))
print("Maximum is:",min(marks))
marks.sort()
print("Marks are:",marks)
marks.reverse()
print("in reverse:",marks)
```

```python
m1=[90,98,80]
m2=[98,89,90]
m1.extend(m2) #storing both m1 and m2 in m1
print(m1)
```

List comprehension

```python
# rule: new_list=[expression for item in iterable if condition]
n=10
even=[i for i in range(1,n+1) if i%2==0]
print(even)
sq=[x**2 for x in range(1,n+1)]
print(sq)
```
