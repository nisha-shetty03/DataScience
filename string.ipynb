```python
#string indexing and slicing
s="Dark chocolate"
#+ve index
print(s[5])
#-ve index
print(s[-5])
print(s[-len(s)])
print(s[-1])
```

```python
#slicing
s="Darkchocolate"
print(s[::-1])
print(s[4:9:1])
print(s[9::])
```
