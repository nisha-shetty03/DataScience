Column transformation in pandas

```python
import pandas as pd
```

Rows and column selection

```python
sdata=pd.read_csv("subdata_pandas/student.csv")
sdata
```

```python
# To select columns 
sdata[['firstName','fees','age','marks']]

# To select details based on column name withe condition
sdata.loc[sdata.firstName=="Nisha"]
sdata.loc[sdata.fees>=30000]

# Slicing on the rows
sdata.loc[1:5]  # 1=starting index, 5= ending index. i.e 1 to 5 index records will be displayed

#Slicing on both rows and columns
sdata.iloc[:4,:3]
sdata.iloc[2:7,1:4] # similar to slicing 2=start row 7-1=end row , 1=start col 4-1=end col
```

Filter

```python
sdata['age']>=20        # boolean
std_age_filter=sdata[sdata['age']>20]  #values
std_age_filter
```

```python
sdata.where(sdata['age']>20,"Minor") 
sdata.where(sdata['age']>20,other="Minor")
sdata.loc[(sdata['marks']>=85,"result")]="selected"
sdata.loc[(sdata['marks']<85,"result")]="Not_selected"
sdata
```

```python
sdata["fullname"]=sdata["firstName"]+" "+sdata["lastName"]
sdata
```
