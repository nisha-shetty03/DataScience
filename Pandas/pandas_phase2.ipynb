Dealing with duplicate values missing data

```python
import pandas as pd
import numpy as np
```

```python
cdata=pd.read_csv("subdata_pandas/company.csv")
cdata
```

Duplicate values

```python
#Overall duplicte
cdata.duplicated()          #boolean value
cdata[cdata.duplicated()]   #select row
#Based on
cdata['id'].duplicated()    #op in boolean
cdata[cdata['id'].duplicated()] 
cdata.drop_duplicates()         # Overall duplicate delection(considering entire column matching)
cdata['id']=cdata['id'].drop_duplicates()   # based on employee id 
```

Missing data

```python
# Checking null values in the small dataset 
cdata.isnull()

# Checking null values in the larger dataset 
cdata.isnull().sum()
'''
op:
id                      0
gender                  0
dateofbirth             1
education               0
department              0
salary                  0
starting_salary         0
job_time                0
previous_experience    25
minority                3
dtype: int64
'''
cdata['id'].isnull()

# Eliminate the Nan values
cdata.dropna()

# replace Nan with any values
cdata.replace(np.nan,"not_available") # op: 0 1	Male 03-02-1952	15	IT	57000 27000	8.17 not_available	No

# replacing nan with 3 in previous experiance
cdata['previous_experience']=cdata['previous_experience'].replace(np.nan,3)
cdata
# Replace nan salary with avg
mean=np.mean(cdata['salary'])
mean

#replacing mean by average
# method 1
cdata['salary']=cdata['salary'].replace(np.nan,34419)
cdata

#method 2
cdata['salary']=cdata['salary'].fillna(mean)
cdata
```

```python
# Forward fill
cdata.ffill() #ex: 1 n 3 ; 2 Nan 4 then op: 1 n 3 ; 2 n 4

#Backward fill
cdata.bfill() #x: 1 n 3 ; 2 Nan 4 ; 3 i 5 then op: 1 n 3 ; 2 i 4 ; 3 i 5

cdata.bfill(inplace=True)   # only in workspace not in og file 

cdata['salary']=cdata['salary'].bfill() # to change perticular field 
```
