GroupBy

```python
exp=pd.read_excel("subdata_pandas/expense-2.xlsx")
exp.head(5)
```

```python
# find the total expense for each ccategory
exp.groupby("Category")['Amount'].sum()
exp.groupby("Category")['Amount'].agg(['sum'])

# Give a complete expense summary for each category
exp.groupby("Category")['Amount'].agg(['sum','count','mean','max','min'])

#How much money was paid by each payment mode
exp.groupby("Payment Mode")['Amount'].agg(['sum'])
'''	        
Payment Mode	sum
Card	       17780
Cash	       4944
UPI	           7425
'''

#How many transaction were made by each payment mode 
exp.groupby("Payment Mode")['Amount'].agg(['sum','count'])

#For each payment mode calculate the total amount spent on each category
exp.groupby(["Payment Mode","Category"])['Amount'].agg(['sum',"count"])

#Which category has the highest total expense
exp.groupby('Category')['Amount'].sum().sort_values(ascending=False)

# highest amount spent on each category 
e=exp.groupby(["Category"])['Amount'].agg(['max']) # to print highest amount spent on each category in descending order
e.max()         #to print highest max amount spent among all the categories

```

MergeOrJoin

```python
cust=pd.read_csv("subdata_pandas/customer.csv")
cust
payment=pd.read_csv("subdata_pandas/payment.csv")
payment
```

```python
#Inner join : matching column from left and right
pd.merge(left=cust,right=payment,on='customerid',how='inner') 

#Left join : all values from left table matching values from rigtht table
pd.merge(left=cust,right=payment,on='customerid',how='left')

#right join : all values from right table matching values from left table
pd.merge(left=cust,right=payment,on='customerid',how='right')


#----task------
#cross join : all values from left table matching values from rigth table
pd.merge(left=cust,right=payment,on='customerid',how='')
```
