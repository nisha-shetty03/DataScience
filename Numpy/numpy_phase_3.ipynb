Aggregate functions

```python
import numpy as np
```

```python
arr=np.array([10,20,30,40,50])
print("Sum:",np.sum(arr))
print("Max:",np.max(arr))
print("Min:",np.min(arr))
print("count:",arr.size)
print("cumulitive sum:",np.cumsum(arr))
print("cumulitive prod:",np.cumprod(arr))   
```

Statistical function

```python
import statistics as st
sales=np.array([200,180,150,250,270,300,250,230,500])
print("Mean",np.mean(sales))
print("Median",np.median(sales))
print("Mode",st.mode(sales))    # numpy dosen't have mode so use statistics 
print("Std_div",np.std(sales))  # standard deviation
```

```python
price=np.array([120,180,200,210,220,300])
sales=np.array([15,30,10,12,9,5])
print(np.corrcoef([price,sales])) # corrcoef= corelation of coefficient

# price increase sales decrease then corrcoef=-ve
#price increase sales increase then corrcoef=+ve
```
