Types

```python
#Type 1 : using parentheses

num=(1,2,3,4,5,6)
col=("black","navy","gold")
mix=("rani",2,3.5,True)
nested=(1,(2,3,4),"rani")

#Type 2 : without parentheses

fruits="orange","grapes","mango"
print(type(fruits))

#Type 3 : using tuple() cons

values=((1,2,3,4,5,6))
lis=[1,2,3,4,5,6]
list_tuple=tuple(lis)

#Type 4 : single value

tup=(45,)
print(type(tup))  # single value in tuple

tup=(45)
print(type(tup))  # without comma at the end denotes a integer type not tuple
```

```python
# checking immutability

colo=("black","red","navy","gold")
#colo[1]="silver"  shows TypeError: 'tuple' object does not support item assignment

# Making changes in tuple

lis=list(colo)  #convert to list first
lis[1]="silver" # insert or change the values 
colo=tuple(lis) # convert back into tuple 
print(colo)
```

Tuple packing and unpacking

```python
val1="rani"
val2=2
val3="cse"
#Packing
student=val1,val2,val3
#Unpacking
name,rank,dep=student

print(name)
print(rank)
print(dep)
```

Tuple methods and functions

```python
nums=(10,30,20,40,50,60,60)
# Max
print(max(nums))
# Min
print(min(nums))
# Sorting
sort_nums=sorted(nums)
print(sort_nums)
# Counting 
print("count",nums.count(60))
# Indexing
print("Index",nums.index(20))
```
