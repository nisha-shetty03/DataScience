```python
# if stmt
i=int(input("enter a number"))
if(i==1):
    print("its ok")

# if else
n=3
if(n<2):
    print("hi")
else:
    print("hello")

# if elif else
n=10
if(n>1 and n<5):
    print("less than 5")
elif(n==5):
    print("exactly 5")
else:
    print("greater than 5")
```

```python
# one line if else
n=9
print("greater than 5") if(n>5) else print("less than 5")

#one line value assignment
a=9
b=10
bigger=b if(a>b) else b
print(bigger)
```
